```python
def process_obj(
    in_path, out_path, scale=1 / 1000.0, flip_axes=("y", "z"), reverse_winding=False
):
    """
    Process an OBJ file:
      - scale geometry
      - flip selected axes to preserve RH system
      - optionally reverse face winding
    """
    with open(in_path, "r", encoding="utf-8", errors="ignore") as fin, open(
        out_path, "w", encoding="utf-8"
    ) as fout:
        for line in fin:
            if line.startswith("v "):  # vertex positions
                _, x, y, z, *rest = line.strip().split()
                x, y, z = float(x), float(y), float(z)
                if "x" in flip_axes:
                    x = -x
                if "y" in flip_axes:
                    y = -y
                if "z" in flip_axes:
                    z = -z
                fout.write(
                    f"v {x*scale:.9f} {y*scale:.9f} {z*scale:.9f}"
                    + (" " + " ".join(rest) if rest else "")
                    + "\n"
                )

            elif line.startswith("vn "):  # vertex normals
                _, nx, ny, nz, *rest = line.strip().split()
                nx, ny, nz = float(nx), float(ny), float(nz)
                if "x" in flip_axes:
                    nx = -nx
                if "y" in flip_axes:
                    ny = -ny
                if "z" in flip_axes:
                    nz = -nz
                fout.write(
                    f"vn {nx:.9f} {ny:.9f} {nz:.9f}"
                    + (" " + " ".join(rest) if rest else "")
                    + "\n"
                )

            elif reverse_winding and line.startswith("f "):
                parts = line.strip().split()
                fout.write("f " + " ".join(reversed(parts[1:])) + "\n")

            else:
                fout.write(line)
```

```python
# Example usage in your notebook:
input_file = "textured_mesh.obj"
output_file = "textured_mesh_edited.obj"

# Flip y and z (keeps RH system) and scale by 1/1000
process_obj(input_file, output_file, scale=1 / 1000.0, flip_axes=("y", "z"))
```
